```python
df_test = df_test.merge(df_pred[['date', 'search_id', 'vues_predites']], on=['date', 'search_id'], how='inner')
```

```python
df_test['date'].min(), df_test['date'].max()
```

```python
df_test_agg = df_test.groupby('search_id').agg({'vues_lr': 'sum', 'vues_predites': 'sum'}).reset_index()
```

```python
df_test_agg
```

```python
top_k_search_id_reel = df_test_agg.sort_values(by='vues_lr', ascending=False).head(1_000)['search_id'].tolist()
top_k_search_id_pred = df_test_agg.sort_values(by='vues_predites', ascending=False).head(1_000)['search_id'].tolist()
```

```python
df_test_agg["is_top_k_reel"] = df_test_agg["search_id"].apply(lambda x: x in top_k_search_id_reel)
df_test_agg["is_top_k_pred"] = df_test_agg["search_id"].apply(lambda x: x in top_k_search_id_pred)
```

```python
from sklearn.metrics import recall_score
recall = recall_score(df_test_agg["is_top_k_reel"], df_test_agg["is_top_k_pred"])
print(f"Recall: {recall:.2f}")
```
